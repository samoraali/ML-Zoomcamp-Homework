Data Preprocessing and Feature Engineering

```python
def preprocess_data(df_dataset, df_symptom_severity, df_symptom_description, df_symptom_precaution):

    df = df_dataset.copy()

    symptom_weight_map = dict(zip(df_symptom_severity['Symptom'], df_symptom_severity['weight']))

    severity_columns = []
    for i, symptom_col in enumerate(symptom_columns, 1):
        df[f'severity_{i}'] = df[symptom_col].map(symptom_weight_map).fillna(0)
        severity_columns.append(f'severity_{i}')

    df['total_severity'] = df[severity_columns].sum(axis=1)

    description_map = dict(zip(df_symptom_description['Disease'], df_symptom_description['Description']))
    df['disease_description'] = df['Disease'].map(description_map)

    precaution_columns = []
    for i in range(1, 5):
        col_name = f'Precaution_{i}'
        precaution_columns.append(col_name)
        df[col_name] = df['Disease'].map(df_symptom_precaution.set_index('Disease')[col_name])

    
    df['symptom_count'] = df[symptom_columns].notnull().sum(axis=1)

    return df
```

Data Preprocessing

```python
df_processed = preprocess_data(df_dataset, df_symptom_severity, df_symptom_description, df_symptom_precaution)

print("Processed Dataset info:")
print(df_processed.info())
print("\nSample of Processed Data:")
print(df_processed[['Disease', 'symptom_count', 'total_severity', 'disease_description']].head())
```

```python
def create_feature_matrix(df_processed):
    symptom_cols = [col for col in df_processed.columns if col.startswith('Symptom_')]

    X = df_processed[symptom_cols].notnull().astype(int)

    severity_cols = [col for col in df_processed.columns if col.startswith('severity_')]
    X_severity = df_processed[severity_cols]

    X_combined = pd.concat([X, X_severity], axis = 1)

    y = df_processed['Disease']

    return X_combined, y
```
